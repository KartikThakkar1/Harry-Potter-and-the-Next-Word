```python
import pandas as pd
import numpy as np
import os
```

```python
# go to 'datasets' folder and access all the files

project_folder_path = os.getcwd()
dataset_folder_path = os.path.join(project_folder_path + "\datasets")
```

```python
all_movies_df = []

for i,file in enumerate(os.listdir(dataset_folder_path)):
    
    if i<8:
        
        file_path  = os.path.join(dataset_folder_path,file)
        df = pd.read_csv(file_path)
        df['part'] = i+1
        all_movies_df.append(df)

df_movies = pd.concat(all_movies_df,ignore_index=True)
```

Cleaning:
 - remove rows where dialogues that are less than 4 words -done
 - for each movie character, find out how many times do they appear in the data, if it is not significant, remove the rows corresponding to them -done
 - DO NOT remove punctuations for BERT
 - use a BERT (uncased) tokenizer model
 - use BERT

```python
df_movies['dialog'] = df_movies['dialog'].astype('str')
df_movies['dialog_len'] = df_movies['dialog'].apply(lambda x: len(x.split()))
```

```python
df2 = df_movies[df_movies['dialog_len']>=5]
actors_count = df2['character'].value_counts()
df2 = df2[df2['character'].isin(actors_count[actors_count >= 10].index)]
```

```python
df2.to_csv("prepared_data.csv",index=False)
```
